```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
orig_df = pd.read_csv('thailand_population_2025.csv')
orig_df
```

# To derive the same graph, there are multiple ways

```python
year_arr = np.arange(2550,2566,2)
year_arr
```

```python
plt.plot(year_arr, orig_df['North'])
plt.show()
```

```python
plt.plot(year_arr, orig_df['North'], marker='o', linestyle='-', color='b')
plt.show()
```

```python
plt.plot(year_arr, orig_df['North'], 'o-b')
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(year_arr, orig_df['North'], marker='o', linestyle='-', color='b')
plt.show()
```

```python
fig, ax = plt.subplots()

temp_line, = ax.plot(year_arr, orig_df['North'], label='North')
temp_line.set_marker('o')
temp_line.set_linestyle('-')
temp_line.set_color('b')
plt.show()
```

# Axes usage example

```python
region_arr = ['North', 'Northeast', 'Central', 'South']
color_arr = ['b', 'r', 'g', 'm']
marker_arr = ['o', '*', '^', 's']
line_style_arr = ['-', '--', '-.', ':']

fig, ax = plt.subplots(1,4)
for i in range(4):
  temp_line, = ax[i].plot(year_arr, orig_df[region_arr[i]], label=region_arr[i])
  temp_line.set_marker(marker_arr[i])
  temp_line.set_linestyle(line_style_arr[i])
  temp_line.set_color(color_arr[i])
plt.show()
```

```python
region_arr = ['North', 'Northeast', 'Central', 'South']
color_arr = ['b', 'r', 'g', 'm']
marker_arr = ['o', '*', '^', 's']
line_style_arr = ['-', '--', '-.', ':']

fig, ax = plt.subplots(1,1)
for i in range(4):
  temp_line, = ax.plot(year_arr, orig_df[region_arr[i]], label=region_arr[i])
  temp_line.set_marker(marker_arr[i])
  temp_line.set_linestyle(line_style_arr[i])
  temp_line.set_color(color_arr[i])

ax.legend(loc='upper right')
plt.show()
```

# Graph Components

## Title, Axis labels and Scope

```python
fig, ax = plt.subplots()

temp_line, = ax.plot(year_arr, orig_df['North'], label='North')
temp_line.set_marker('o')
temp_line.set_linestyle('-')
temp_line.set_color('b')

ax.set_title("North Population between 2550-2564", loc='left')
ax.set_xlabel("Year")
ax.set_ylabel("Population")

ax.set_xlim(2550, 2564)
ax.set_ylim(11500000, 12500000)
plt.show()
```

## Ticks, Tick labels, and Grid

```python
fig, ax = plt.subplots()

temp_line, = ax.plot(year_arr, orig_df['North'], label='North')
temp_line.set_marker('o')
temp_line.set_linestyle('-')
temp_line.set_color('b')

ax.set_title("North Population between 2550-2564", loc='left')
ax.set_xlabel("Year")
ax.set_ylabel("Population (millions)")

xticks = np.arange(2550, 2570, 4)
xtick_labels = []
for tick in xticks:
  xtick_labels.append(str(tick))
ax.set_xticks(xticks, xtick_labels)

yticks = np.arange(11500000,12600000,100000)
ytick_vals = np.linspace(1.15,1.25,11)
ytick_labels = []
for tick_val in ytick_vals:
  ytick_labels.append(str(tick_val))
ax.set_yticks(yticks, ytick_labels)

ax.set_xlim(2550, 2564)
ax.set_ylim(11500000, 12500000)

ax.grid()
plt.show()
```

## Detailed Parameters

```python
fig, ax = plt.subplots()

temp_line, = ax.plot(year_arr, orig_df['North'], label='North')
temp_line.set_marker('o')
temp_line.set_linestyle('-')
temp_line.set_color('b')

ax.set_title("North Population between 2550-2564", loc='center')
ax.set_xlabel("Year")
ax.set_ylabel("Population (millions)")

xticks = np.arange(2550, 2570, 4)
xtick_labels = []
for tick in xticks:
  xtick_labels.append(str(tick))
ax.set_xticks(xticks, xtick_labels)

yticks = np.arange(11500000,12600000,100000)
ytick_vals = np.linspace(1.15,1.25,11)
ytick_labels = []
for tick_val in ytick_vals:
  ytick_labels.append(str(tick_val))
ax.set_yticks(yticks, ytick_labels)

ax.set_xlim(2550, 2564)
ax.set_ylim(11500000, 12500000)

ax.grid()

ax.tick_params(axis='x', direction='inout', length=8, width=5, colors='purple',
               grid_color='g', grid_alpha=0.4, grid_linestyle='--',
               labeltop=True, labelrotation=45, labelcolor='c', labelsize=8)

ax.tick_params(axis='y', direction='out', length=10, width=2, colors='r', right=True,
               grid_color='m', grid_alpha=1.0, grid_linewidth=1, grid_linestyle='-.',
               labelright=True, labelrotation=90, labelcolor='orange', labelsize=12)
plt.show()
```
