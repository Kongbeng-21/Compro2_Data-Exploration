```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
orig_df = pd.read_csv('thailand_population_2025_unsorted.csv')
orig_df
```

```python
# Set Year to be index of dataframe

df = orig_df.set_index('Year').copy()
df
```

## Line Graph

```python
# Construct x-axis that is time array

year_arr = np.arange(2550,2566,2)
year_arr
```

```python
# Matplotlib 
# Line example 1

region_arr = df.columns
color_arr = ['b', 'g', 'r', 'y', 'm']
marker_arr = ['o', '*', '^', 's', 'p']
line_style_arr = ['-', '--', '-.', ':', '--']

fig, ax = plt.subplots(figsize = (8,6))
for i in range(len(region_arr)):
    temp_line, = ax.plot(year_arr, df[region_arr[i]], label=region_arr[i])
    temp_line.set_marker(marker_arr[i])
    temp_line.set_linestyle(line_style_arr[i])
    temp_line.set_color(color_arr[i])

    
ax.set_xlim(2550, 2564)
xticks = np.arange(2550, 2566, 2)
xtick_labels = [str(tick) for tick in xticks]
ax.set_xticks(year_arr)
ax.set_xticklabels(xtick_labels, rotation=0)

ax.set_ylim(0,25000000)

ax.legend(loc='upper center', ncols=5)

ax.set_xlabel('Year')
ax.set_ylabel('Population')
ax.set_title('Thailand Population between 2550-2564')
plt.show()
```

## Bar Graph

```python
# Construct x-axis that is time array

year_arr = np.arange(2550, 2566, 2)
year_arr
```

```python
# Matplotlib
# Bar example 1

fig, ax = plt.subplots()
ax.bar(year_arr, df['Bangkok'])
ax.set_xlabel('Year')
ax.set_ylabel('Population (million)')
ax.set_title('Bangkok population between 2550-2564')
plt.show()
```

```python
# Matplotlib
# Bar example 2

fig, ax = plt.subplots()
ax.bar(year_arr, df['Bangkok'], color='Blue')

ax.set_xticks(year_arr)
xtick_labels = [str(tick) for tick in year_arr]
ax.set_xticklabels(xtick_labels, rotation=45)

ax.set_ylim(0,7000000)
yticks = np.arange(0, 8000000, 1000000)
ax.set_yticks(yticks)
ytick_values = np.arange(0, 8, 1)
ytick_labels = [str(tick) for tick in ytick_values]
ax.set_yticklabels(ytick_labels)

ax.set_xlabel('Year')
ax.set_ylabel('Population (million)')
ax.set_title('Bangkok population between 2550-2564')
```

```python
# Seaborn

plt.figure(figsize = (8,6))

ax = sns.barplot(x=year_arr, y=df['Bangkok'], color='b')

ax.set_xlabel('Year')
ax.set_ylabel('Population')
ax.set_title('Bangkok Population between 2550-2564')
```

```python
# Seaborn
# Bar example 1

plt.figure(figsize = (8,6))

ax = sns.barplot(x=year_arr, y=df['Bangkok'], color='b')

ax.set_xticklabels(year_arr)

ax.set_ylim(0,7000000)
yticks = np.arange(0, 8000000, 1000000)
ytick_values = np.linspace(0, 7, 8)
ytick_labels = [f'{tick:.0f}'.format(tick) for tick in ytick_values]
ax.set_yticks(yticks)
ax.set_yticklabels(ytick_labels)

ax.set_xlabel('Year')
ax.set_ylabel('Population (million)')
ax.set_title('Bangkok Population between 2550-2564')
```

## Multiple Bar Graph

```python
# Current dataframe
df
```

```python
# Sort the data based on the most recent year from the most population to the least population
sorted_df = df.sort_values(by=2564, axis=1, ascending=False)
sorted_df
```

```python
# Create new dataframe that have 3 columns: Year, Region, Population
stacked_df = sorted_df.stack().reset_index()
stacked_df.head()
```

```python
# Rename columns in the new dataframe
stacked_df = stacked_df.rename(columns={'level_1':'Region', 0:'Population'})
stacked_df.head()
```

```python
# Seaborn
# Multiple bar graph example 1

plt.figure(figsize = (8,6))

ax = sns.barplot(x='Year', y='Population', data=stacked_df, hue='Region')

ax.set_ylim(0, 25000000)
ax.legend(loc='upper center', ncols=5)

ax.set_xlabel('Year')
ax.set_ylabel('Thailand Population')
ax.set_title('Thailand Population between 2550-2564')
```

```python
# Seaborn
# Multiple bar graph example 2

plt.figure(figsize = (8,6))

ax = sns.barplot(x='Region', y='Population', data=stacked_df, hue='Year')

ax.set_xlabel('Region')
ax.set_ylabel('Thailand Population')
ax.set_title('Thailand Population between 2550-2564')
```
