```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
orig_df = pd.read_csv('thailand_population_2025_unsorted.csv')
orig_df
```

```python
# Set column Year to be index of dataframe

df = orig_df.set_index('Year').copy()
df
```

## Pie Graph

### Help links:

##### Pie 
https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.pie.html

##### Circle
https://matplotlib.org/stable/api/_as_gen/matplotlib.patches.Circle.html

##### Example with more complicated labels
https://matplotlib.org/stable/gallery/pie_and_polar_charts/pie_and_donut_labels.html#sphx-glr-gallery-pie-and-polar-charts-pie-and-donut-labels-py

```python
# Use the data from the most recent year (the last row) to plot
# Sort regions with the most population to the least population

y = df.iloc[len(df)-1].sort_values(ascending=False)  
y
```

```python
# Get regions

region_arr = y.index
region_arr
```

```python
# Matplotlib
# Pie graph example 1

fig, ax = plt.subplots()
ax.pie(y)
ax.set_title('Thailand Population in 2564')
plt.show()
```

```python
# Matplotlib
# Pie graph example 2

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots()
ax.pie(y, 
       labels=region_arr, 
       colors=color_arr,
       startangle=90, counterclock=False,
       autopct='%1.1f%%')
ax.axis('equal')
ax.set_title('Thailand Population in 2564')
plt.show()
```

```python
# Matplotlib
# Pie graph example 3

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots()
slices, texts, numbers = ax.pie(y, 
                                colors=color_arr,
                                startangle=90, counterclock=False,
                                autopct='%1.2f%%',
                                textprops={'color':'w'})
ax.set_title('Thailand Population in 2564')
ax.legend(slices, 
          region_arr,
          title="Regions",
          bbox_to_anchor=(1, 1))
plt.show()
```

```python
# Matplotlib
# Pie graph example 4

color_arr = ['b', 'g', 'r', 'y', 'm']
explosion_factor = [0.02, 0.04, 0.06, 0.08, 0.1]

fig, ax = plt.subplots()
ax.pie(y, 
       labels=region_arr, 
       colors=color_arr,
       startangle=90, counterclock=False,
       autopct='%1.1f%%',
       explode=explosion_factor)
ax.set_title('Thailand Population in 2564')
plt.show()
```

```python
# Matplotlib
# Pie graph example 5

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots()
ax.pie(y, 
       labels=region_arr, 
       colors=color_arr,
       startangle=90, counterclock=False,
       autopct='%1.1f%%', pctdistance=1.15, 
       labeldistance=1.25, rotatelabels=True,
       radius=1.5  # set radius of pie graph
       )
ax.set_title('Thailand Population in 2564', pad=100.0)
plt.show()
```

```python
# Matplotlib
# Donut graph example 

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots()
ax.pie(y, 
       labels=region_arr, 
       colors=color_arr,
       startangle=90, counterclock=False,
       autopct='%1.1f%%', pctdistance=0.75)

# Create a white circle and add to the axes
white_circle_radius = 0.50    # radius value can be changed
central_circle = plt.Circle((0, 0), white_circle_radius, fc='white')
ax.add_patch(central_circle)

ax.set_title('Thailand Population in 2564')
plt.show()
```

## Bar Graph

### Help links

##### Matplotlib : Bar
https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.bar.html

###### Seaborn : Bar
https://seaborn.pydata.org/generated/seaborn.barplot.html

```python
# Use the data from the most recent year (the last row) to plot
# Sort regions with the most population to the least population
# Also, get regions

y = df.iloc[len(df)-1].sort_values(ascending=False)  
region_arr = y.index
y
```

```python
# Matplotlib
# Bar graph example 1

fig, ax = plt.subplots()
ax.bar(region_arr, y)

ax.set_xlabel('Region')
ax.set_ylabel('Population')
ax.set_title('Thailand Population in 2564')
plt.show()
```

```python
# Matplotlib
# Bar graph example 2

fig, ax = plt.subplots(figsize=(6, 4))

# a list of hex color codes can be assigned to bar color.
color_list = ['#3e1d7d', '#f6784f', '#3c86c2', '#f580e0', '#fffea3']
ax.bar(region_arr, y, color=color_list, edgecolor='r', linewidth=2)

ax.set_xticks(region_arr)
ax.set_xticklabels(region_arr, rotation=45)

ax.set_ylim(0,25000000)
yticks = np.arange(0, 30000000, 5000000)
ytick_values = np.linspace(0, 25, 6)
ytick_labels = [f'{tick:.0f}'.format(tick) for tick in ytick_values]
ax.set_yticks(yticks)
ax.set_yticklabels(ytick_labels)

ax.set_xlabel('Region')
ax.set_ylabel('Population (million)')
ax.set_title('Thailand Population in 2564')
plt.show()
```

```python
# Seaborn
# Bar graph example 1

plt.figure(figsize = (6,4))

color_list = ['#3e1d7d', '#f6784f', '#3c86c2', '#f580e0', '#fffea3']
ax = sns.barplot(x=region_arr, y=y, hue=region_arr, palette=color_list, legend=False)

ax.set_xlabel('Region')
ax.set_ylabel('Population')
ax.set_title('Thailand Population in 2564')
plt.show()
```

```python
# Seaborn
# Bar graph example 2

plt.figure(figsize = (6,4))

# Seaborn also  has built-in color palette available.  Read more at: https://r02b.github.io/seaborn_palettes/
ax = sns.barplot(x=region_arr, y=y, hue=region_arr, palette="bright", legend=False)

xticks = np.arange(0, len(region_arr))
ax.set_xticks(xticks, region_arr)
ax.tick_params(axis='x', rotation=45)

ax.set_ylim(0,25000000)
yticks = np.arange(0, 30000000, 5000000)
ytick_values = np.linspace(0, 25, 6)
ytick_labels = [f'{tick:.0f}'.format(tick) for tick in ytick_values]
ax.set_yticks(yticks, ytick_labels)

#leg = plt.legend()
#ax.get_legend().remove()

ax.set_xlabel('Region')
ax.set_ylabel('Population (million)')
ax.set_title('Thailand Population in 2564')
plt.show()
```

## Stacked Bar Graph

### Help link

##### Matplotlib : Stacked bar chart
https://matplotlib.org/stable/gallery/lines_bars_and_markers/bar_stacked.html

```python
# Sort the data based on the most recent year from the most population to the least population
sorted_df = df.sort_values(by=2564, axis=1, ascending=False)
sorted_df
```

```python
# Also, get regions
region_arr = sorted_df.columns
region_arr
```

```python
# Also, get years
year_arr = sorted_df.index
year_arr
```

```python
# Matplotlib
# Stacked bar example 1

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots()
bottom = np.zeros(len(year_arr))

for i in range(len(region_arr)):
    ax.bar(year_arr, 
           sorted_df[region_arr[i]], 
           label=region_arr[i], 
           color=color_arr[i], 
           bottom=bottom)
    bottom += sorted_df[region_arr[i]]
    
ax.legend(loc='upper right', ncols=3)
ax.set_ylim(0,100000000)

ax.set_xlabel('Year')
ax.set_ylabel('Population')
ax.set_title('Thailand Population between 2550-2564')
plt.show()
```

```python
# Matplotlib
# Stacked bar example 2

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots(figsize = (8,6))

bottom = np.zeros(len(year_arr))
for i in range(len(region_arr)):
    ax.bar(year_arr, 
           sorted_df[region_arr[i]], 
           label=region_arr[i], 
           color=color_arr[i], 
           bottom=bottom)
    bottom += sorted_df[region_arr[i]]
    
ax.set_xticks(year_arr)
ax.set_xticklabels(year_arr)

ax.set_ylim(0,80000000)
yticks = np.arange(0, 80000000, 10000000)
ytick_values = np.linspace(0, 70, 8)
ytick_labels = [f'{tick:.0f}'.format(tick) for tick in ytick_values]
ax.set_yticks(yticks)
ax.set_yticklabels(ytick_labels)

ax.legend(loc='upper center', ncols=5)

ax.set_xlabel('Year')
ax.set_ylabel('Population (million)')
ax.set_title('Thailand Population between 2550-2564')
plt.show()
```

```python
# Seaborn
# Stacked bar example 1

plt.figure(figsize = (8,6))

bottom = np.zeros(len(year_arr))
for i in range(len(region_arr)):
    ax = sns.barplot(x=year_arr, y=sorted_df[region_arr[i]], label=region_arr[i], 
               color=color_arr[i], bottom=bottom)
    bottom += sorted_df[region_arr[i]]
        
xticks = np.arange(0, len(year_arr))
ax.set_xticks(xticks, year_arr)

ax.set_ylim(0,80000000)
yticks = np.arange(0, 80000000, 10000000)
ytick_values = np.linspace(0, 70, 8)
ytick_labels = [f'{tick:.0f}'.format(tick) for tick in ytick_values]
ax.set_yticks(yticks)
ax.set_yticklabels(ytick_labels)

ax.legend(loc='upper center', ncols=5)

ax.set_xlabel('Year')
ax.set_ylabel('Population (million)')
ax.set_title('Thailand Population between 2550-2564')
plt.show()
```

## Stacked Area Graph

##### Matplotlib : Line 
https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.plot.html

##### Matplotlib : Fill the area between lines
https://matplotlib.org/stable/gallery/lines_bars_and_markers/fill_between_demo.html

```python
# Sort the data based on the most recent year from the most population to the least population
sorted_df = df.sort_values(by=2564, axis=1, ascending=False)
sorted_df
```

```python
# Get regions
region_arr = sorted_df.columns
region_arr
```

```python
# Matplotlib
# Stacked area example 1

color_arr = ['b', 'g', 'r', 'y', 'm']

fig, ax = plt.subplots(1,1)

cumulative = np.zeros(len(year_arr))
for i in range(len(region_arr)):
    last_cumulative = cumulative.copy()
    cumulative += sorted_df[region_arr[i]]
    ax.plot(year_arr, cumulative, label=region_arr[i], color=color_arr[i])
    ax.fill_between(year_arr, cumulative, last_cumulative, color=color_arr[i], alpha=0.3)
        
ax.legend(loc='upper right', ncols=3)
ax.set_ylim(0,100000000)

ax.set_xlabel('Year')
ax.set_ylabel('Population')
ax.set_title('Thailand Population between 2550-2564')
plt.show()
```
